```python
import PyPDF2
import docx

def extract_text_from_file(file_path):
    text = ""
    if file_path.endswith(".txt"):
        with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
            text = f.read()
    elif file_path.endswith(".pdf"):
        reader = PyPDF2.PdfReader(file_path)
        for page in reader.pages:
            text += page.extract_text() + " "
    elif file_path.endswith(".docx"):
        doc = docx.Document(file_path)
        for para in doc.paragraphs:
            text += para.text + " "
    else:
        text = "Unsupported file format."
    return text.strip()


def predict(input_text, input_file):
    if input_text:
        result = predict_with_full_info(input_text)
    elif input_file:
        extracted_text = extract_text_from_file(input_file)
        if extracted_text.strip():
            result = predict_with_full_info(extracted_text)
        else:
            return "Could not extract text from the uploaded file.", "", "", "", ""
    else:
        return "Please provide text or document.", "", "", "", ""

    return (
        result["Predicted Age Group"],
        result["Advice"],
        result["Eligibility ✅"],
        result["Not Eligible ❌"],
        result["Healthcare Tips 💊"]
    )
```
